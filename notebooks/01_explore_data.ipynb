```python
import pandas as pd
import numpy as np
import os
import glob

print("Libraries loaded ✅")
```

```python
data_folder = "../data/files/"
all_files = sorted(glob.glob(data_folder + "*.txt"))

print("Total files found:", len(all_files))
print("First 3 files:", all_files[:3])
```

```python
sample = pd.read_csv(all_files[0], sep='\t')

print("Columns:", sample.columns.tolist())
sample.head()
```

```python
dfs = []

for filepath in all_files:
    filename = os.path.basename(filepath)
    user_id = filename.split("_")[0]
    
    df_temp = pd.read_csv(
        filepath,
        sep='\t',
        encoding='latin1',
        on_bad_lines='skip',
        engine='python'
    )
    
    df_temp["PARTICIPANT_ID"] = int(user_id)
    
    dfs.append(df_temp)

df = pd.concat(dfs, ignore_index=True)
df.to_csv("../data/keystroke_data.csv", index=False)

print("✅ CSV file created successfully!")

print("Total rows:", len(df))
print("Total users:", df["PARTICIPANT_ID"].nunique())
```
